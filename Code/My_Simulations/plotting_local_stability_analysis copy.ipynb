```python
import os
import pandas as pd
import numpy as np
import json as json
import seaborn as sns

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
```

```python
def parse_and_label_locally_stable_strategies(df):


    df["policy"] = df["policy"].apply(lambda x: np.array(json.loads(x)))
    df = df.explode("policy", ignore_index=True)


    df[["strategy_agent_1", "strategy_agent_2"]] = pd.DataFrame([[p[0], p[1]] for p in df["policy"]])


    df["strategy_agent_1_label"] = df.apply(lambda row: strategy_to_label(row["strategy_agent_1"], mode = row["mode"]), axis=1)
    df["strategy"] = df["strategy_agent_1_label"]

    df.drop(columns=["strategy_agent_1", "strategy_agent_2", "policy", "strategy_agent_1_label"], inplace=True)
    df.rename(columns={"mode": "info"}, inplace=True)
    mapping = {
    "ecological": "Only Ecological",
    "social": "Only Social",
    "complete": "Complete",
    "none": "No Info"
    }
    df["info"] = df["info"].replace(mapping)
    return df

```

```python
def plot_locally_stable_strategies(df, m_values_to_plot, discount_factors_to_plot):

    df_for_plotting = df[(df["m_value"].isin(m_values_to_plot)) & (df["discount_factor"].isin(discount_factors_to_plot))].copy()

    desired_order = [ "ALL D",  "Alternate C/D",  "WSLS","ALL C",]  

    strategies_to_plot = [s for s in desired_order if s in df_for_plotting["strategy"].unique()]
    
    y_index = {s:i for i,s in enumerate(strategies_to_plot)}

    y_jitter = {'complete': 0.25, 'ecological': 0.08,
                'social': -0.08, 'none': -0.25}

    df_for_plotting["y"] = df_for_plotting["strategy"].map(y_index)
    df_for_plotting["y_jit"] = df_for_plotting["y"] + df_for_plotting["mode"].map(y_jitter)

   
    sns.set_theme(context="talk", style="white")
    plt.rcParams.update({
        "axes.facecolor": "white",
        "figure.facecolor": "white",
        "axes.edgecolor":  "black",
        "grid.color":      "#ffffff",
        "grid.linestyle":  "--",
        "grid.linewidth":  0.6,
        "xtick.color":     "black",
        "ytick.color":     "black",
        "text.color":      "black",
        "axes.labelcolor": "black",
    })

    fig, ax = plt.subplots(figsize=(9, 5))

    # Strategy edge colors
    strategy_colors = {
        "ALL D": "#d62728", 
        "ALL C": "#2ca02c",  
        "WSLS": "#2ca02c",   # slightly darker green
        "Alternate C/D": "#8b3e4c"
    }

    # Marker shapes for info conditions
    marker_map = {
        "social": "^", 
        "ecological": "s", 
        "complete": "o", 
        "none": "X"
    }

    # --------------------------
    # 4. Scatterplot
    # --------------------------
    for mode in all_information_modes:
        sub = df_for_plotting[df_for_plotting["mode"] == mode]
        if sub.empty:
            continue
        sns.scatterplot(
            data=sub, x="discount_factor", y="y_jit",
            marker=marker_map[mode], s=150, linewidth=1.8,
            facecolors="none",
            edgecolor=sub["strategy"].map(strategy_colors),
            ax=ax, legend=False
        )

    # --------------------------
    # 5. Legend (outside)
    # --------------------------
    shape_handles = [
        Line2D([0],[0], marker=marker_map[i], linestyle="None",
            markerfacecolor="none", markeredgecolor="black",
            markersize=9, label=i)
        for i in all_information_modes
    ]

    ax.legend(
        handles=shape_handles,
        title="Information condition",
        loc="center left",
        bbox_to_anchor=(1.02, 0.5),
        frameon=True, facecolor="white", edgecolor="black"
    )

    # --------------------------
    # 6. Axis formatting
    # --------------------------
    ax.set_xlabel("Discount factor")
    ax.set_ylabel("Strategy")

    # Force ticks at your discount factor values
    ax.set_xticks(discount_factors_to_plot)
    ax.set_xticklabels([f"{d:.3f}" for d in discount_factors_to_plot])

    ax.set_yticks(range(len(strategies_to_plot)))
    ax.set_yticklabels(strategies_to_plot)

    # Gridlines along x
    # ax.yaxis.grid(True)
    # ax.xaxis.grid(True)
    # ax.xaxis.grid(True, linestyle="--", alpha=0.7, color="#aaaaaa")
    midpoints = [(discount_factors_to_plot[i] + discount_factors_to_plot[i+1]) / 2
                for i in range(len(discount_factors_to_plot)-1)]

    # for xpos in midpoints:
    #     ax.axvline(
    #         x=xpos, linestyle="-", color="#aaaaaa",
    #         linewidth=0.8, alpha=1
    #     )


    sns.despine()
    plt.tight_layout()
    plt.show()
```

```python

def plot_locally_stable_strategies(df_parsed, m_values_to_plot, discount_factors, strategies):
    df = df_parsed.copy()

    df = df[df["m_value"].isin(m_values_to_plot)].copy()
    info_conditions = [ "No Info", "Only Social", "Only Ecological", "Complete"]

    # ordered categories & sort
    cat_info = pd.CategoricalDtype(info_conditions, ordered=True)
    cat_strat = pd.CategoricalDtype(strategies, ordered=True)
    df["info"] = df["info"].astype(cat_info)
    df["strategy"] = df["strategy"].astype(cat_strat)
    df = df.sort_values(["info", "discount_factor", "strategy"]).reset_index(drop=True)

    sns.set_theme(context="talk", style="whitegrid")
    plt.rcParams.update({
        "axes.facecolor": "white",
        "figure.facecolor": "white",
        "axes.edgecolor":  "black",
        "grid.color":      "#ffffff",
        "grid.linestyle":  "--",
        "grid.linewidth":  1,
        "xtick.color":     "black",
        "ytick.color":     "black",
        "text.color":      "black",
        "axes.labelcolor": "black",
    })

    # Base y by info row
    df["y"] = df["info"].cat.codes.astype(float)

    # --- Dynamic vertical jitter per (info, discount_factor) ---
    def offsets_for_strats(strat_list):
        """Return centered offsets for the given ordered strategy list."""
        n = len(strat_list)
        if n <= 1:
            return {strat_list[0]: 0.0} if n == 1 else {}
        # Evenly spaced, centered around 0
        center = (n - 1) / 2
        spacing = 0.17
        pos = [(i - center) * spacing for i in range(n)]
        return {s: p for s, p in zip(strat_list, pos)}  #function that decides how much offset to give for each strategy based on the discount factor and info condition

    def assign_group_offsets(g):
        # Keep deterministic order using the master 'strategies' list
        present_in_order = [s for s in strategies if s in g["strategy"].unique()] #makes sure the strategy is in the order i define globally
        offs = offsets_for_strats(present_in_order) #simply gives offsets to the stratgy, which is a dictionry (strat:offset)
        return g["strategy"].map(offs).astype(float)   #

    df["y_jit_delta"] = (
        df
        .groupby(["info", "discount_factor"], group_keys=False, observed = True) #group keys = false to avoid cluttering
        .apply(assign_group_offsets, include_groups=False) #apply the offset function to each group of info and discount factor
        .fillna(0.0) #offset 0 if no strategy is present
        .astype(float)
    )
    df["y_jit"] = df["y"] + df["y_jit_delta"]

    # Markers, colors (your existing maps)
    marker_map = {
        "ALL D": "X",
        "ALL C": "o",
        "WSLS": "^",
        "Alternate C/D": "s",
        "GT": "P",
    }
    strategy_colors = {
        "ALL D": "#d62728",
        "ALL C": "#1f77b4",   # blue
        "WSLS": "#1f77b4",    # blue
        "Alternate C/D": "#c7324d",
        "GT": "#1f77b4",      # blue
    }

    # Plot
    fig, ax = plt.subplots(figsize=(9.5, 4.5))
    for strat in strategies:
        sub = df[df["strategy"] == strat]
        if sub.empty:
            continue
        sns.scatterplot(
            data=sub,
            x="discount_factor",
            y="y_jit",
            marker=marker_map.get(strat, "o"),
            s=150,
            linewidth=1.8,
            facecolors="none",
            edgecolor=strategy_colors.get(strat, "black"),
            ax=ax,
            legend=False
        )

    # Strategy legend
    shape_handles = [
    Line2D([0],[0],
           marker=marker_map.get(s, "o"),
           linestyle="None",
           markerfacecolor="none",
           markeredgecolor=strategy_colors.get(s, "black"),
           markersize=9,
           label=strategy_display.get(s, s)   # <--- CHANGE HERE
    )
    for s in strategies
]

    ax.legend(
        handles=shape_handles,
        title="Strategy",
        loc="center left",
        bbox_to_anchor=(1.02, 0.5),
        frameon=True,
        facecolor="white",
        edgecolor="black"
    )

    # Axes & ticks
    # ax.set_xlabel(rf"Discount factor ($\delta$)")
    ax.set_ylabel("Information condition")
    # ax.yaxis.set_ticks_position("left")
    # ax.tick_params(
    #     axis="y",
    #     which="major",
    #     length=6,
    #     width=1.2,
    #     color="black",
    #     left=True
    # )
    ax.set_xticks(discount_factors)

    # ax.set_xticklabels([f"{d:.3f}" for d in discount_factors])


    ax.set_xlabel("Foresight (discount factor)")

    ax.set_xticks(discount_factors)

    ax.set_xticklabels([
        "Low",      # δ = 0.955
        "",         # δ = 0.965
        "Intermed.",   # δ = 0.975
        "",         # δ = 0.985
        "High"      # δ = 0.995
    ])


    base_y_positions = df.groupby("info", observed=False)["y"].mean().values

    ax.set_yticks(base_y_positions)
    ax.set_yticklabels(info_conditions)
    # ax.tick_params(axis = 'y', rotation = 30)
    ax.minorticks_on()                               # enable minor ticks
    ax.tick_params(
    axis="x",
    which="major",
    length=6,      # visible tick size
    width=1.2,
    bottom=True,   # force ticks ON
    top=False,
    color="black"
)



    # ax.yaxis.grid(True)
    # ax.xaxis.grid(True)
    sns.despine()
    plt.tight_layout()
    plt.title(f"Locally stable strategies")

    fig.text(
        0.90, 0.25, 
        s =  rf"$m = {m_values_to_plot[0]}$",
        transform=fig.transFigure, 
        ha="right", va="top",
        fontsize=11,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, boxstyle="round,pad=0.3")
    )

    fig.text(
        0.90, 0.89, 
        s =  rf"$\gamma$",
        transform=fig.transFigure, 
        ha="center", va="top",
        fontsize=10,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, boxstyle="round,pad=0.3")
    )
    fig.text(
        0.90, 0.85, 
        s =  rf"${0.995, 0.965, 0.975, 0.985, 0.995}$",
        transform=fig.transFigure, 
        ha="center", va="top",
        fontsize=10,
        bbox=dict(facecolor="white", edgecolor="none", alpha=0.7, boxstyle="round,pad=0.3")
    )
   

    #set slight grey face color
    ax.set_facecolor("#ffffff")


    plt.show()
    # fig.savefig(rf"C:\Users\localuser\Desktop\Social_Ecological_Info_Cooperation\Code\Plots\Thesis\Local_Stabilty_Analysis\local_stability_analysis_m_6.5.png", dpi=300, bbox_inches="tight")
    # fig.savefig(rf"d:\Personal Projects\Social_Ecological_Info_Cooperation\Code\Plots\Thesis\Local_Stabilty_Analysis\local_stability_analysis_m_6.5.png", dpi=300, bbox_inches="tight")



```
