```python
import numpy as np
import matplotlib.pyplot as plt

# --- Prospect Theory parameters (make effect more visible) ---
alpha = 0.6   # curvature for gains -> smaller = more flattening
beta = 0.6    # curvature for losses
lambda_ = 2.5 # strong loss aversion

# --- Prospect Theory value function ---
def prospect_value(x):
    return np.where(x >= 0, x**alpha, -lambda_ * ((-x)**beta))

# --- Generate x values ---
x = np.linspace(-10, 10, 400)
y = prospect_value(x)

# --- Plot ---
plt.figure(figsize=(8, 6))
plt.plot(x, y, color='royalblue', linewidth=3, label='Prospect Theory Value Function')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Add a few reference points to illustrate diminishing sensitivity
for val in [1, 2, 4, 8]:
    plt.scatter(val, prospect_value(val), color='green', s=60)
    plt.scatter(-val, prospect_value(-val), color='red', s=60)
    plt.text(val+0.3, prospect_value(val), f"{prospect_value(val):.2f}", color='green')
    plt.text(-val-3, prospect_value(-val), f"{prospect_value(-val):.2f}", color='red')

plt.title("Prospect Theory Value Curve\n(Diminishing Sensitivity + Loss Aversion)", fontsize=14)
plt.xlabel("Outcome (x)", fontsize=12)
plt.ylabel("Subjective Value v(x)", fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
